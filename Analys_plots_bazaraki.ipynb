```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('transform_lots.csv')
```

```python
df.head()
df.info()
df.describe()
df.isnull().sum()
```

Introduction to numerical data Plot area, price fo meter, coverage and density, how much day ago ads was posted

```python
meter_price=df['Square meter price'].dropna()
```

```python
sns.histplot(df['Square meter price'], bins=100)
#plt.xlim(0, 700000)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Plot Type', y='Square meter price')

plt.title('price per meter on plot type')
plt.xlabel('plot type')
plt.ylabel('Price per m2')
plt.xticks(rotation=45)
plt.ylim(0, 4000)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='City', y='Square meter price')

plt.title('price per meter on city')
plt.xlabel('city')
plt.ylabel('Price per m2')
plt.xticks(rotation=45)
plt.ylim(0, 4000)
plt.show()
```
